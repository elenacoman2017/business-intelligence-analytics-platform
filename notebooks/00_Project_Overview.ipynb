# Business Intelligence and Predictive Analytics Platform

## 1. Project Introduction

### 1.1 Background

Organisations generate and collect large amounts of data through their daily business activities. This information can provide valuable insights into sales performance, customer behaviour, financial activities, operational efficiency and potential risks.

However, collecting data does not automatically lead to better business decisions. Data must first be organised, validated, analysed and interpreted to identify meaningful patterns and trends.

Business intelligence and data science provide methods for transforming raw data into useful information. Descriptive analytics can explain historical performance, statistical analysis can investigate relationships between variables, and machine learning can identify patterns and generate predictions.

This project aims to demonstrate how these techniques can be combined within a single analytical platform.

### 1.2 Project Aim

The aim of this project is to develop a Python-based Business Intelligence and Predictive Analytics Platform that applies statistical analysis, data visualisation and machine learning techniques to different business scenarios.

The platform will cover six analytical areas:

1. Sales analytics.
2. Customer analytics.
3. Financial analytics.
4. Fraud and anomaly detection.
5. Operational analytics.
6. Predictive analytics and forecasting.

### 1.3 Project Objectives

The main objectives are:

- Develop a structured data processing pipeline.
- Investigate and prepare datasets from different business scenarios.
- Apply exploratory data analysis and statistical techniques.
- Implement and evaluate appropriate machine learning algorithms.
- Identify unusual patterns and potential anomalies.
- Develop predictive models for selected business problems.
- Incorporate explainability techniques where appropriate.
- Communicate analytical findings through visualisations and reports.
- Follow good software engineering practices, including modular programming, testing, documentation and version control.

### 1.4 Project Scope

The project will use publicly available and synthetic datasets. It will not use confidential employer information or real personal financial or medical records.

The initial development will take place using Python and Jupyter Notebook. Reusable Python components will be developed separately to support future application development.

The project is intended as a technical portfolio demonstration. Its results will not be presented as validated production systems or as substitutes for professional financial, fraud or operational decisions.

### 1.5 Development Approach

The project will follow the CRISP-DM methodology:

- Business understanding.
- Data understanding.
- Data preparation.
- Modelling.
- Evaluation.
- Deployment.

The methodology will provide a structured approach to the analytical work while allowing individual stages to be revisited when necessary.

### 1.6 Expected Outcome

The expected outcome is a documented and reproducible analytical project demonstrating the application of Python, statistical analysis, machine learning, data visualisation and software engineering techniques across several business scenarios.

````python
from pathlib import Path

# ============================================================
# PURPOSE:
# Create a .gitignore file in the main project directory.
#
# WHAT IS .gitignore?
# Git uses this file to identify files and folders that should
# not be included in version control or uploaded to GitHub.
#
# WHY IS IT IMPORTANT?
# It helps prevent accidental publication of private datasets,
# passwords, API keys, temporary files and unnecessary files.
# ============================================================

# Define the main project directory.
# This is the location containing the notebooks, data, src,
# tests and requirements.txt folders/files.
project_dir = Path(
    r"C:\Users\elena\Business Intelligence and Predictive Analytics Platform"
)

# Define the rules for files and folders that Git should ignore.
# Each line represents a file pattern or folder to exclude.
gitignore_content = """
# ------------------------------------------------------------
# PYTHON TEMPORARY FILES
# ------------------------------------------------------------

# Python creates __pycache__ folders when code is executed.
# These contain temporary compiled files and do not need to
# be uploaded to GitHub.
__pycache__/
*.py[cod]

# ------------------------------------------------------------
# JUPYTER NOTEBOOK TEMPORARY FILES
# ------------------------------------------------------------

# Jupyter automatically creates checkpoint folders.
# These are temporary backups of notebook files.
.ipynb_checkpoints/

# ------------------------------------------------------------
# PRIVATE DATASETS
# ------------------------------------------------------------

# Ignore the contents of the raw data folder.
# Raw datasets may contain confidential or licensed information.
data/raw/*

# Ignore the contents of the processed data folder.
# Processed datasets can also contain sensitive information.
data/processed/*

# Ignore the contents of the external data folder.
# External datasets may have redistribution restrictions.
data/external/*

# Keep placeholder files so Git preserves empty data folders.
!data/raw/.gitkeep
!data/processed/.gitkeep
!data/external/.gitkeep

# ------------------------------------------------------------
# ENVIRONMENT AND CONFIGURATION FILES
# ------------------------------------------------------------

# Local environment variables may contain API keys or secrets.
.env
.env.*

# Allow an example environment file to be published.
!.env.example

# Ignore local virtual environments.
.venv/
venv/
env/

# ------------------------------------------------------------
# OPERATING SYSTEM FILES
# ------------------------------------------------------------

# Windows may create these automatically.
Thumbs.db
Desktop.ini

# macOS may create these automatically.
.DS_Store

# ------------------------------------------------------------
# GENERATED OUTPUTS
# ------------------------------------------------------------

# Ignore generated figures and reports by default.
# We can selectively include approved portfolio examples later.
outputs/figures/*
outputs/reports/*

# Keep placeholder files so Git preserves these folders.
!outputs/figures/.gitkeep
!outputs/reports/.gitkeep

# ------------------------------------------------------------
# EDITOR AND DEVELOPMENT FILES
# ------------------------------------------------------------

# Ignore common editor settings and temporary files.
.vscode/
.idea/
*.log
"""

# Create the .gitignore file in the main project directory.
gitignore_path = project_dir / ".gitignore"

# Save the contents using UTF-8 encoding.
gitignore_path.write_text(
    gitignore_content.strip() + "\n",
    encoding="utf-8"
)

# Confirm that the file has been created.
print("SUCCESS: .gitignore created.")

# Display the exact file location.
print("\nFile location:")
print(gitignore_path)

# Display the saved rules so they can be checked.
print("\nFile contents:")
print(gitignore_path.read_text(encoding="utf-8"))
````

````python
from pathlib import Path

# ============================================================
# PURPOSE:
# Create .gitkeep files inside folders that are currently empty.
#
# WHY IS THIS NECESSARY?
# Git tracks files, not empty folders.
# A .gitkeep file allows GitHub to display these folders even
# when we have not added any datasets, figures or reports yet.
# ============================================================

# Locate the main project directory.
project_dir = Path(
    r"C:\Users\elena\Business Intelligence and Predictive Analytics Platform"
)

# List the folders that need placeholder files.
folders_to_preserve = [
    "data/raw",
    "data/processed",
    "data/external",
    "outputs/figures",
    "outputs/reports"
]

# Create one .gitkeep file in each folder.
for folder in folders_to_preserve:

    # Build the full path to the folder.
    folder_path = project_dir / folder

    # Create the folder if it does not already exist.
    folder_path.mkdir(parents=True, exist_ok=True)

    # Create the placeholder file.
    placeholder_path = folder_path / ".gitkeep"

    # Write a short explanation inside the placeholder file.
    placeholder_path.write_text(
        "This file keeps the folder visible in Git.\n",
        encoding="utf-8"
    )

    # Confirm which placeholder was created.
    print(f"Created: {placeholder_path.relative_to(project_dir)}")

print("\nAll placeholder files created successfully.")
````

````python

# ============================================================
# PURPOSE:
# Inspect the Git repository before the first commit.
#
# WHAT THIS CHECKS:
# 1. Confirms that the project is a Git repository.
# 2. Checks whether .gitignore exists.
# 3. Lists files Git considers untracked or modified.
# 4. Checks that private data and generated outputs are ignored.
#
# IMPORTANT:
# This cell does not upload or commit anything.
# ============================================================

from pathlib import Path
import subprocess

# ------------------------------------------------------------
# STEP 1: DEFINE THE PROJECT LOCATION
# ------------------------------------------------------------

project_dir = Path(
    r"C:\Users\elena\Business Intelligence and Predictive Analytics Platform"
)

if not (project_dir / ".git").exists():
    raise RuntimeError("Git repository not found.")

# ------------------------------------------------------------
# STEP 2: CREATE A REUSABLE GIT COMMAND FUNCTION
# ------------------------------------------------------------

def run_git(command):
    """Run a Git command inside the project folder."""

    result = subprocess.run(
        ["git"] + command,
        cwd=project_dir,
        capture_output=True,
        text=True
    )

    if result.returncode != 0:
        raise RuntimeError(result.stderr.strip())

    return result.stdout.strip()

# ------------------------------------------------------------
# STEP 3: CHECK THE .gitignore FILE
# ------------------------------------------------------------

gitignore_path = project_dir / ".gitignore"

print("GITIGNORE CHECK")
print("----------------")

if gitignore_path.exists():
    print("PASS: .gitignore exists.")

    gitignore_content = gitignore_path.read_text(
        encoding="utf-8"
    )

    print("File size:", gitignore_path.stat().st_size, "bytes")

else:
    raise FileNotFoundError(".gitignore is missing.")

# ------------------------------------------------------------
# STEP 4: INSPECT PROJECT FILE STATUS
# ------------------------------------------------------------

print("\nPROJECT FILE STATUS")
print("-------------------")

# --short provides a concise list of changed and untracked files.
# --untracked-files=all displays individual untracked files.

file_status = run_git([
    "status",
    "--short",
    "--untracked-files=all"
])

if file_status:
    print(file_status)
else:
    print("No untracked or modified files found.")

# ------------------------------------------------------------
# STEP 5: CHECK IMPORTANT IGNORE RULES
# ------------------------------------------------------------

print("\nIGNORE RULE VERIFICATION")
print("------------------------")

# These are representative paths.
# Git checks the matching rules even if the files do not exist.

paths_to_check = [
    "data/raw/example.csv",
    "data/processed/example.csv",
    "data/external/example.csv",
    "outputs/figures/example.png",
    "outputs/reports/example.pdf",
    ".env",
    "__pycache__/example.pyc",
    "notebooks/.ipynb_checkpoints/example.ipynb"
]

for path in paths_to_check:

    result = subprocess.run(
        ["git", "check-ignore", "-v", path],
        cwd=project_dir,
        capture_output=True,
        text=True
    )

    if result.returncode == 0:
        print("PASS: Ignored:", path)
    else:
        print("CHECK: Not ignored:", path)

# ------------------------------------------------------------
# STEP 6: CHECK THE CURRENT BRANCH
# ------------------------------------------------------------

print("\nGIT BRANCH")
print("----------")

branch = run_git(["branch", "--show-current"])

print("Current branch:", branch)

print("\nInspection completed.")
print("No files have been committed or uploaded.")
````

````python
# ============================================================
# PURPOSE:
# Compare the two notebook files to identify whether one
# is an accidental duplicate.
#
# WHAT THIS CHECKS:
# 1. Whether both files exist.
# 2. Their file sizes.
# 3. Their modification dates.
# 4. Whether their contents are identical.
#
# No files are changed or deleted.
# ============================================================

from pathlib import Path
import hashlib
from datetime import datetime

# ------------------------------------------------------------
# STEP 1: DEFINE BOTH FILE LOCATIONS
# ------------------------------------------------------------

project_dir = Path(
    r"C:\Users\elena\Business Intelligence and Predictive Analytics Platform"
)

notebook_files = [
    project_dir / "00_Project_Overview.ipynb.ipynb",
    project_dir / "notebooks" / "00_Project_Overview.ipynb"
]

# ------------------------------------------------------------
# STEP 2: INSPECT EACH FILE
# ------------------------------------------------------------

file_hashes = {}

for file_path in notebook_files:

    print("\nFile:", file_path.name)
    print("Location:", file_path.parent)

    if file_path.exists():

        print("Exists: YES")
        print("Size:", file_path.stat().st_size, "bytes")

        modified = datetime.fromtimestamp(
            file_path.stat().st_mtime
        )

        print("Last modified:", modified)

        # SHA-256 creates a fingerprint of the file contents.
        # Identical fingerprints mean identical file contents.

        file_hash = hashlib.sha256(
            file_path.read_bytes()
        ).hexdigest()

        file_hashes[str(file_path)] = file_hash

        print("SHA-256:", file_hash)

    else:
        print("Exists: NO")

# ------------------------------------------------------------
# STEP 3: COMPARE THE FILES
# ------------------------------------------------------------

print("\nFILE COMPARISON")
print("----------------")

if len(file_hashes) == 2:

    hashes = list(file_hashes.values())

    if hashes[0] == hashes[1]:
        print("The two notebook files are identical.")

    else:
        print("The two notebook files are different.")

else:
    print("Both files were not found, so comparison is incomplete.")

print("\nInspection completed. No files were changed.")
````
