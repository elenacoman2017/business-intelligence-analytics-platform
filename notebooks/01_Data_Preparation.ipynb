### 7.5 Testing CSV and Excel Loading

The data loader will be tested using a small synthetic business sales dataset.

Synthetic data is used to ensure that the testing process does not expose real customer information or confidential business records.

The test will:

1. Create a fictional sales dataset.
2. Save the dataset in CSV and Excel formats.
3. Load both files using `load_dataset()`.
4. Verify the number of records and columns.
5. Check that the loaded data matches the original dataset.

These checks establish whether the data loader can consistently import both supported file formats without performing unintended data transformations.

```python
# ============================================================
# TEST CSV AND EXCEL DATA LOADING
# ============================================================
#
# PURPOSE:
# Verify that the reusable data loader can correctly load
# CSV and Excel files.
#
# WHY THIS IS NEEDED:
# Importing a function does not prove that it works.
# We must test its actual file-loading functionality.
#
# REPRODUCIBILITY:
# The fictional sales dataset is generated by this code.
# No external dataset is required.
#
# ============================================================

import pandas as pd
from pathlib import Path

# ------------------------------------------------------------
# STEP 1: CREATE THE TEST DATA DIRECTORY
# ------------------------------------------------------------

# Store the synthetic files in data/raw/.
# These local test files are excluded from Git.

raw_data_dir = project_dir / "data" / "raw"
raw_data_dir.mkdir(parents=True, exist_ok=True)

print("PASS: Test data directory is available.")

# ------------------------------------------------------------
# STEP 2: CREATE A SYNTHETIC SALES DATASET
# ------------------------------------------------------------

# All records are fictional and created for testing purposes.

sample_sales = pd.DataFrame({
    "order_id": [1001, 1002, 1003, 1004, 1005],
    "order_date": [
        "2026-01-05",
        "2026-01-06",
        "2026-01-07",
        "2026-01-08",
        "2026-01-09"
    ],
    "customer_id": [201, 202, 203, 201, 204],
    "product": [
        "Laptop",
        "Mouse",
        "Keyboard",
        "Monitor",
        "Headphones"
    ],
    "quantity": [1, 2, 1, 1, 3],
    "unit_price": [850.00, 25.00, 45.00, 180.00, 35.00],
    "region": [
        "London",
        "Birmingham",
        "Manchester",
        "London",
        "Bristol"
    ]
})

# Calculate revenue using quantity multiplied by unit price.

sample_sales["revenue"] = (
    sample_sales["quantity"] * sample_sales["unit_price"]
)

print("PASS: Synthetic sales dataset created.")

# ------------------------------------------------------------
# STEP 3: SAVE THE DATASET IN CSV AND EXCEL FORMATS
# ------------------------------------------------------------

csv_file = raw_data_dir / "sample_sales.csv"
excel_file = raw_data_dir / "sample_sales.xlsx"

sample_sales.to_csv(csv_file, index=False)

sample_sales.to_excel(
    excel_file,
    index=False,
    sheet_name="Sales"
)

print("PASS: CSV test file created.")
print("PASS: Excel test file created.")

# ------------------------------------------------------------
# STEP 4: LOAD THE CSV FILE
# ------------------------------------------------------------

csv_data = load_dataset(csv_file)

print("\nCSV LOADING TEST")
print("=" * 50)
print(f"Rows loaded: {len(csv_data)}")
print(f"Columns loaded: {len(csv_data.columns)}")

# ------------------------------------------------------------
# STEP 5: LOAD THE EXCEL FILE
# ------------------------------------------------------------

excel_data = load_dataset(
    excel_file,
    sheet_name="Sales"
)

print("\nEXCEL LOADING TEST")
print("=" * 50)
print(f"Rows loaded: {len(excel_data)}")
print(f"Columns loaded: {len(excel_data.columns)}")

# ------------------------------------------------------------
# STEP 6: VERIFY THE DATASET STRUCTURE
# ------------------------------------------------------------

expected_rows = 5
expected_columns = 8

assert len(csv_data) == expected_rows
assert len(excel_data) == expected_rows

assert len(csv_data.columns) == expected_columns
assert len(excel_data.columns) == expected_columns

assert list(csv_data.columns) == list(sample_sales.columns)
assert list(excel_data.columns) == list(sample_sales.columns)

print("\nPASS: CSV structure verified.")
print("PASS: Excel structure verified.")

# ------------------------------------------------------------
# STEP 7: STANDARDISE DATES AND COMPARE THE DATASETS
# ------------------------------------------------------------

# Convert both date columns to the same datetime format.
# Excel and CSV readers may interpret dates differently.

csv_data["order_date"] = pd.to_datetime(
    csv_data["order_date"]
)

excel_data["order_date"] = pd.to_datetime(
    excel_data["order_date"]
)

# Compare the underlying data without requiring identical
# inferred data types across the two file formats.

pd.testing.assert_frame_equal(
    csv_data,
    excel_data,
    check_dtype=False
)

print("PASS: CSV and Excel datasets contain matching data.")

# ------------------------------------------------------------
# STEP 8: DISPLAY THE LOADED DATA
# ------------------------------------------------------------

print("\nLOADED SAMPLE SALES DATA")
display(csv_data)

print("\nDATA LOADING TESTS COMPLETED SUCCESSFULLY.")
```

### 7.6 Testing Error Handling

The data loader must identify invalid inputs and raise appropriate Python exceptions rather than allowing unsuitable files or paths to proceed through the pipeline.

Three error scenarios will be tested:

1. **Missing file:** the specified file does not exist.
2. **Invalid path:** the supplied path refers to a directory rather than a file.
3. **Unsupported format:** the supplied file uses an extension that the data loader does not support.

The expected exceptions are:

- `FileNotFoundError` for a missing file.
- `IsADirectoryError` for a directory path.
- `ValueError` for an unsupported file format.

These checks help identify input problems at the earliest stage of data preparation.

```python
import tempfile
from pathlib import Path

# Create a temporary directory for the error-handling tests.
# Temporary files will be removed automatically when testing finishes.
with tempfile.TemporaryDirectory() as temporary_directory:

    test_dir = Path(temporary_directory)

    # ------------------------------------------------------------
    # TEST 1: MISSING FILE
    # ------------------------------------------------------------
    # Attempt to load a CSV file that does not exist.

    missing_file = test_dir / "missing.csv"

    try:
        load_dataset(missing_file)

    except FileNotFoundError:
        print("PASS: Missing files raise FileNotFoundError.")

    else:
        raise AssertionError(
            "Expected FileNotFoundError for a missing file."
        )

    # ------------------------------------------------------------
    # TEST 2: DIRECTORY PATH
    # ------------------------------------------------------------
    # Attempt to load a directory instead of a dataset file.

    try:
        load_dataset(test_dir)

    except IsADirectoryError:
        print("PASS: Directory paths raise IsADirectoryError.")

    else:
        raise AssertionError(
            "Expected IsADirectoryError for a directory path."
        )

    # ------------------------------------------------------------
    # TEST 3: UNSUPPORTED FILE FORMAT
    # ------------------------------------------------------------
    # Create a temporary text file and attempt to load it.

    unsupported_file = test_dir / "sample.txt"

    unsupported_file.write_text(
        "This file is used to test unsupported formats.",
        encoding="utf-8"
    )

    try:
        load_dataset(unsupported_file)

    except ValueError:
        print("PASS: Unsupported formats raise ValueError.")

    else:
        raise AssertionError(
            "Expected ValueError for an unsupported file format."
        )

print("\nERROR-HANDLING TESTS COMPLETED SUCCESSFULLY.")
```

```python
# ------------------------------------------------------------
# SECTION 1: IDENTIFY COMPLETE DUPLICATE RECORDS
# ------------------------------------------------------------
# duplicated() identifies rows that match an earlier row
# across all columns.
#
# keep="first" means the first occurrence is not marked
# as a duplicate, but subsequent identical rows are.

duplicate_mask = sales_data.duplicated(keep="first")

# ------------------------------------------------------------
# SECTION 2: COUNT DUPLICATE RECORDS
# ------------------------------------------------------------
# Sum the Boolean results to calculate the number of duplicates.

duplicate_count = int(duplicate_mask.sum())

print("DUPLICATE RECORD ASSESSMENT")
print("=" * 65)
print(f"Total records: {len(sales_data)}")
print(f"Duplicate records: {duplicate_count}")

# ------------------------------------------------------------
# SECTION 3: DISPLAY DUPLICATE RECORDS
# ------------------------------------------------------------
# Display matching records only when duplicates are identified.

print("\nIDENTIFIED DUPLICATE RECORDS")
print("=" * 65)

if duplicate_count > 0:

    duplicate_records = sales_data.loc[duplicate_mask]

    display(duplicate_records)

else:

    print("No complete duplicate records identified.")

# ------------------------------------------------------------
# SECTION 4: VERIFY THE ASSESSMENT
# ------------------------------------------------------------
# Confirm that the duplicate count is within the valid range.

if not 0 <= duplicate_count < len(sales_data):
    raise AssertionError(
        "The duplicate record count is outside the expected range."
    )

print("\nPASS: Duplicate record assessment completed.")
print("PASS: Duplicate records checked across all columns.")

print("\nDUPLICATE RECORD ASSESSMENT COMPLETED.")
```

### 10.6 Categorical Consistency Assessment

Categorical consistency refers to the standardised representation of values within descriptive columns.

Inconsistent entries can occur when the same category is recorded using different spellings, capitalisation, spacing or abbreviations.

For example, the values `London`, `london` and ` London ` may represent the same geographical region but would be treated as separate categories by many analytical operations.

The assessment will examine the `product` and `region` columns for:

- Leading or trailing whitespace.
- Inconsistent capitalisation.
- Empty or whitespace-only strings.
- The number of distinct recorded categories.

The original values will not be modified during this assessment. Any identified inconsistencies will be documented and addressed separately during data cleaning, if appropriate.

```python
# ------------------------------------------------------------
# SECTION 1: DEFINE CATEGORICAL COLUMNS
# ------------------------------------------------------------
# These columns contain descriptive category values.

categorical_columns = ["product", "region"]

# ------------------------------------------------------------
# SECTION 2: ASSESS CATEGORICAL CONSISTENCY
# ------------------------------------------------------------
# Examine each column without modifying its original values.

consistency_results = []

for column in categorical_columns:

    values = sales_data[column].dropna().astype(str)

    # Identify values containing leading or trailing whitespace.
    whitespace_count = int(
        values.ne(values.str.strip()).sum()
    )

    # Identify values containing lowercase characters.
    lowercase_count = int(
        values.ne(values.str.upper())
        .where(values.ne(values.str.lower()), False)
        .sum()
    )

    # Identify empty or whitespace-only strings.
    empty_count = int(
        values.str.strip().eq("").sum()
    )

    # Count the distinct values currently recorded.
    distinct_count = int(values.nunique())

    consistency_results.append({
        "Column": column,
        "Distinct Values": distinct_count,
        "Whitespace Issues": whitespace_count,
        "Lowercase Entries": lowercase_count,
        "Empty Strings": empty_count
    })

# ------------------------------------------------------------
# SECTION 3: DISPLAY ASSESSMENT RESULTS
# ------------------------------------------------------------
# Present the findings in a structured table.

consistency_summary = pd.DataFrame(consistency_results)

print("CATEGORICAL CONSISTENCY ASSESSMENT")
print("=" * 65)

display(consistency_summary)

# ------------------------------------------------------------
# SECTION 4: VERIFY THE ASSESSMENT
# ------------------------------------------------------------
# Confirm that both expected categorical columns were assessed.

if list(consistency_summary["Column"]) != categorical_columns:
    raise AssertionError(
        "Not all expected categorical columns were assessed."
    )

if consistency_summary[
    ["Whitespace Issues", "Lowercase Entries", "Empty Strings"]
].isnull().any().any():
    raise AssertionError(
        "Unexpected missing assessment results were identified."
    )

print("\nPASS: Both categorical columns assessed.")
print("PASS: Whitespace, capitalisation and empty strings checked.")

print("\nCATEGORICAL CONSISTENCY ASSESSMENT COMPLETED.")
```

Date fields are important for time-based business analysis, including sales trends, seasonal patterns and transaction frequency.

The checks will identify:

The assessment will use `pandas.to_datetime()` with error handling to test whether the recorded values can be interpreted as dates.

```python
# ============================================================
# SECTION 10.10: DATE VALUE ASSESSMENT
# ============================================================
# PURPOSE:
# Examine the order_date column to determine whether its
# recorded values can be interpreted as valid dates.
#
# The assessment checks:
# 1. Missing date values.
# 2. Invalid date values.
# 3. Earliest and latest valid dates.
# 4. Number of distinct valid dates.
#
# IMPORTANT:
# The conversion is performed on a temporary Series.
# The original sales_data DataFrame remains unchanged.

# ------------------------------------------------------------
# SECTION 1: VERIFY THE DATE COLUMN
# ------------------------------------------------------------
# Confirm that the expected date column exists.

date_column = "order_date"

if date_column not in sales_data.columns:
    raise KeyError(
        f"Date column not found: {date_column}"
    )

print("PASS: Order date column exists.")

# ------------------------------------------------------------
# SECTION 2: CONVERT VALUES TEMPORARILY
# ------------------------------------------------------------
# errors="coerce" converts unrecognised date values to NaT.
# NaT represents a missing or invalid datetime value.
#
# This conversion does not overwrite the original column.

converted_dates = pd.to_datetime(
    sales_data[date_column],
    errors="coerce"
)

print("PASS: Temporary date conversion completed.")

# ------------------------------------------------------------
# SECTION 3: CALCULATE DATE QUALITY METRICS
# ------------------------------------------------------------
# Count missing values in the original date column.

missing_dates = int(
    sales_data[date_column].isnull().sum()
)

# Count non-empty values that could not be converted to dates.
invalid_dates = int(
    (
        sales_data[date_column].notnull()
        & converted_dates.isnull()
    ).sum()
)

# Count successfully interpreted dates.
valid_dates = int(converted_dates.notnull().sum())

# Calculate the number of distinct valid dates.
distinct_dates = int(converted_dates.nunique())

# Identify the earliest and latest valid dates.
earliest_date = converted_dates.min()
latest_date = converted_dates.max()

# ------------------------------------------------------------
# SECTION 4: DISPLAY THE ASSESSMENT RESULTS
# ------------------------------------------------------------

date_quality_summary = pd.DataFrame({
    "Measure": [
        "Total Records",
        "Missing Dates",
        "Invalid Dates",
        "Valid Dates",
        "Distinct Valid Dates"
    ],
    "Result": [
        len(sales_data),
        missing_dates,
        invalid_dates,
        valid_dates,
        distinct_dates
    ]
})

print("\nDATE VALUE QUALITY ASSESSMENT")
print("=" * 65)

display(date_quality_summary)

print("\nDATE RANGE")
print("=" * 65)
print(f"Earliest date: {earliest_date}")
print(f"Latest date: {latest_date}")

# ------------------------------------------------------------
# SECTION 5: VERIFY THE ASSESSMENT
# ------------------------------------------------------------
# Confirm that all records have been accounted for.

if missing_dates + invalid_dates + valid_dates != len(sales_data):
    raise AssertionError(
        "The date assessment counts do not reconcile."
    )

if valid_dates == 0:
    raise ValueError(
        "No valid dates were identified."
    )

print("\nPASS: Date values assessed.")
print("PASS: Missing and invalid dates checked.")
print("PASS: Date range and distinct dates calculated.")

print("\nDATE VALUE ASSESSMENT COMPLETED.")
```

### 11.13 Unit Price and Revenue Validation

The `unit_price` and `revenue` columns represent important financial measures within the sales dataset.

The following business rules will be applied:

**Unit price**

- Values must not be missing.
- Values must be numerical.
- Values must not be negative.

**Revenue**

- Values must not be missing.
- Values must be numerical.
- Values must not be negative.
- Recorded revenue must equal quantity multiplied by unit price.

The revenue calculation will be checked using the following relationship:

\[
\text{Revenue} = \text{Quantity} \times \text{Unit Price}
\]

A numerical tolerance will be applied when comparing calculated and recorded revenue to account for minor floating-point representation differences.

These rules assume that the synthetic dataset records gross line revenue without separate discounts, taxes or adjustments. More complex financial datasets would require additional business rules.

```python
# ============================================================
# SECTION 11.13: UNIT PRICE AND REVENUE VALIDATION
# ============================================================
# PURPOSE:
# Validate unit prices and recorded revenue against the
# financial business rules defined for this sales dataset.
#
# BUSINESS RULES:
# 1. Unit prices must not be missing or negative.
# 2. Revenue must not be missing or negative.
# 3. Revenue must equal quantity multiplied by unit price.
#
# The original sales_data DataFrame will not be modified.

# ------------------------------------------------------------
# SECTION 1: VERIFY REQUIRED COLUMNS
# ------------------------------------------------------------

financial_columns = [
    "quantity",
    "unit_price",
    "revenue"
]

for column in financial_columns:

    if column not in sales_data.columns:
        raise KeyError(
            f"Required financial column not found: {column}"
        )

    if not pd.api.types.is_numeric_dtype(sales_data[column]):
        raise TypeError(
            f"Column {column} must contain numerical values."
        )

print("PASS: All required financial columns exist.")
print("PASS: All selected financial columns are numerical.")

# ------------------------------------------------------------
# SECTION 2: IDENTIFY MISSING AND NEGATIVE VALUES
# ------------------------------------------------------------

unit_price = sales_data["unit_price"]
revenue = sales_data["revenue"]

missing_unit_price = unit_price.isnull()
negative_unit_price = unit_price.lt(0)

missing_revenue = revenue.isnull()
negative_revenue = revenue.lt(0)

# ------------------------------------------------------------
# SECTION 3: VALIDATE THE REVENUE CALCULATION
# ------------------------------------------------------------
# Calculate expected revenue using quantity multiplied by
# unit price.
#
# The calculation is temporary and does not overwrite revenue.

calculated_revenue = (
    sales_data["quantity"] * unit_price
)

# Compare recorded revenue with calculated revenue.
# A tolerance of 0.01 allows a difference of up to one penny.

revenue_tolerance = 0.01

revenue_difference = (
    revenue - calculated_revenue
).abs()

# Only compare rows where all relevant values are available.
revenue_comparable = (
    sales_data["quantity"].notnull()
    & unit_price.notnull()
    & revenue.notnull()
)

revenue_mismatch = (
    revenue_comparable
    & revenue_difference.gt(revenue_tolerance)
)

# ------------------------------------------------------------
# SECTION 4: COUNT VALIDATION ISSUES
# ------------------------------------------------------------

financial_validation_results = pd.DataFrame({

    "Validation Rule": [
        "Unit price must not be missing",
        "Unit price must not be negative",
        "Revenue must not be missing",
        "Revenue must not be negative",
        "Revenue must match quantity multiplied by unit price"
    ],

    "Issues Identified": [
        int(missing_unit_price.sum()),
        int(negative_unit_price.sum()),
        int(missing_revenue.sum()),
        int(negative_revenue.sum()),
        int(revenue_mismatch.sum())
    ]

})

financial_validation_results["Validation Status"] = (
    financial_validation_results["Issues Identified"]
    .eq(0)
    .map({True: "PASS", False: "FAIL"})
)

print("\nUNIT PRICE AND REVENUE VALIDATION")
print("=" * 75)

display(financial_validation_results)

# ------------------------------------------------------------
# SECTION 5: DISPLAY REVENUE COMPARISON
# ------------------------------------------------------------
# Show recorded and calculated revenue for each transaction.

revenue_comparison = pd.DataFrame({

    "Order ID": sales_data["order_id"],

    "Quantity": sales_data["quantity"],

    "Unit Price": unit_price,

    "Recorded Revenue": revenue,

    "Calculated Revenue": calculated_revenue,

    "Difference": revenue_difference

})

print("\nREVENUE CALCULATION COMPARISON")
print("=" * 75)

display(revenue_comparison)

# ------------------------------------------------------------
# SECTION 6: DISPLAY AFFECTED RECORDS IF NECESSARY
# ------------------------------------------------------------

invalid_financial_mask = (
    missing_unit_price
    | negative_unit_price
    | missing_revenue
    | negative_revenue
    | revenue_mismatch
)

if invalid_financial_mask.any():

    print("\nRecords requiring financial investigation:")

    display(
        sales_data.loc[
            invalid_financial_mask,
            [
                "order_id",
                "quantity",
                "unit_price",
                "revenue"
            ]
        ]
    )

# ------------------------------------------------------------
# SECTION 7: VERIFY VALIDATION RESULTS
# ------------------------------------------------------------

if invalid_financial_mask.any():
    raise AssertionError(
        "Unit price or revenue validation failed. "
        "Investigate the affected records before proceeding."
    )

print("\nPASS: All unit prices are present and non-negative.")
print("PASS: All revenue values are present and non-negative.")
print("PASS: Recorded revenue matches the calculated revenue.")

print("\nUNIT PRICE AND REVENUE VALIDATION COMPLETED.")
```

### 11.15 Product and Region Validation

Product and regional information supports sales analysis, product performance reporting and geographical comparisons.

The validation will examine the `product` and `region` columns.

The following business rules will apply:

- Product values must not be missing.
- Region values must not be missing.
- Product and region values must not contain empty or whitespace-only strings.
- Text values must not contain leading or trailing whitespace.

The validation will also report the distinct products and regions represented in the dataset.

This assessment will not impose a fixed list of acceptable products or regions. Such restrictions would require an approved business reference list.

```python
# ============================================================
# SECTION 11.15: PRODUCT AND REGION VALIDATION
# ============================================================
# PURPOSE:
# Validate the completeness and formatting of product and
# region values used in the sales dataset.
#
# BUSINESS RULES:
# 1. Product and region values must not be missing.
# 2. Text values must not be empty or whitespace-only.
# 3. Text values must not contain leading or trailing spaces.
#
# The original sales_data DataFrame will not be modified.

# ------------------------------------------------------------
# SECTION 1: DEFINE CATEGORICAL COLUMNS
# ------------------------------------------------------------

categorical_columns = [
    "product",
    "region"
]

# ------------------------------------------------------------
# SECTION 2: VERIFY COLUMN AVAILABILITY AND DATA TYPES
# ------------------------------------------------------------

for column in categorical_columns:

    if column not in sales_data.columns:
        raise KeyError(
            f"Required categorical column not found: {column}"
        )

    if not (
        pd.api.types.is_object_dtype(sales_data[column])
        or pd.api.types.is_string_dtype(sales_data[column])
    ):
        raise TypeError(
            f"Column {column} must contain text-compatible values."
        )

print("PASS: Product and region columns exist.")
print("PASS: Both columns contain text-compatible values.")

# ------------------------------------------------------------
# SECTION 3: ASSESS EACH CATEGORICAL COLUMN
# ------------------------------------------------------------

categorical_validation_results = []

categorical_issue_masks = {}

for column in categorical_columns:

    values = sales_data[column]

    # Identify missing values.
    missing_mask = values.isnull()

    # Convert non-missing values to strings for text checks.
    text_values = values.astype("string")

    # Identify empty strings and strings containing only spaces.
    empty_mask = (
        text_values.notnull()
        & text_values.str.strip().eq("").fillna(False)
    )

    # Identify leading or trailing whitespace.
    whitespace_mask = (
        text_values.notnull()
        & text_values.ne(text_values.str.strip()).fillna(False)
    )

    # Store the combined issue mask for this column.
    categorical_issue_masks[column] = (
        missing_mask | empty_mask | whitespace_mask
    )

    # Count the identified issues.
    categorical_validation_results.append({

        "Column": column,

        "Missing Values": int(missing_mask.sum()),

        "Empty Strings": int(empty_mask.sum()),

        "Whitespace Issues": int(whitespace_mask.sum()),

        "Distinct Values": int(values.nunique())

    })

# ------------------------------------------------------------
# SECTION 4: DISPLAY VALIDATION RESULTS
# ------------------------------------------------------------

categorical_validation_summary = pd.DataFrame(
    categorical_validation_results
)

print("\nPRODUCT AND REGION VALIDATION")
print("=" * 75)

display(categorical_validation_summary)

# ------------------------------------------------------------
# SECTION 5: DISPLAY DISTINCT BUSINESS CATEGORIES
# ------------------------------------------------------------

for column in categorical_columns:

    print(f"\nDISTINCT VALUES: {column.upper()}")
    print("=" * 65)

    display(
        sales_data[column]
        .dropna()
        .drop_duplicates()
        .sort_values()
        .reset_index(drop=True)
    )

# ------------------------------------------------------------
# SECTION 6: VERIFY VALIDATION RESULTS
# ------------------------------------------------------------

total_categorical_issues = sum(
    int(mask.sum())
    for mask in categorical_issue_masks.values()
)

if total_categorical_issues > 0:

    print("\nRecords requiring categorical investigation:")

    for column, mask in categorical_issue_masks.items():

        if mask.any():

            print(f"\nAffected records in {column}:")

            display(
                sales_data.loc[
                    mask,
                    ["order_id", column]
                ]
            )

    raise AssertionError(
        "Product or region validation failed. "
        "Investigate the affected records before proceeding."
    )

print("\nPASS: No missing product or region values.")
print("PASS: No empty or whitespace-only values.")
print("PASS: No leading or trailing whitespace detected.")

print("\nPRODUCT AND REGION VALIDATION COMPLETED.")
```

## 11.17 Overall Business Rule Validation Summary

The final validation step consolidates the business rules assessed throughout this section.

The purpose is to verify that the dataset satisfies the defined structural, identifier, date, numerical and categorical requirements before any cleaning or transformation is performed.

The consolidated assessment will report the outcome of each validation rule and identify any failures requiring investigation.

No automatic corrections will be applied during this process. Any identified issues must be investigated before the dataset proceeds to the cleaning stage.

```python
# ============================================================
# SECTION 11.17: OVERALL BUSINESS RULE VALIDATION SUMMARY
# ============================================================
# PURPOSE:
# Independently recheck the main business rules defined in
# Section 11 and produce a consolidated validation report.
#
# IMPORTANT:
# This cell assesses the original sales_data DataFrame.
# It does not modify the original dataset or correct values.

# ------------------------------------------------------------
# SECTION 1: DEFINE REQUIRED COLUMNS
# ------------------------------------------------------------

required_columns = [
    "order_id",
    "order_date",
    "customer_id",
    "product",
    "quantity",
    "unit_price",
    "region",
    "revenue"
]

missing_columns = [
    column
    for column in required_columns
    if column not in sales_data.columns
]

if missing_columns:
    raise KeyError(
        f"Required columns are missing: {missing_columns}"
    )

# ------------------------------------------------------------
# SECTION 2: PREPARE VALIDATION RESULTS
# ------------------------------------------------------------

business_rule_results = []

def record_validation(rule, passed, details):
    """
    Store the outcome of an individual business rule.

    Parameters
    ----------
    rule:
        Name of the validation rule.

    passed:
        True if the rule passed, otherwise False.

    details:
        Short explanation of the validation result.
    """

    business_rule_results.append({
        "Validation Rule": rule,
        "Result": "PASS" if passed else "FAIL",
        "Details": details
    })

# ------------------------------------------------------------
# SECTION 3: VALIDATE IDENTIFIERS
# ------------------------------------------------------------

# Order identifiers must be present and unique.
order_id_valid = (
    sales_data["order_id"].notna().all()
    and not sales_data["order_id"].duplicated().any()
)

record_validation(
    "Order identifiers are present and unique",
    order_id_valid,
    f"{sales_data['order_id'].nunique()} distinct order IDs"
)

# Customer identifiers must be present.
customer_id_valid = sales_data["customer_id"].notna().all()

record_validation(
    "Customer identifiers are present",
    customer_id_valid,
    f"{sales_data['customer_id'].nunique()} distinct customers"
)

# ------------------------------------------------------------
# SECTION 4: VALIDATE TRANSACTION DATES
# ------------------------------------------------------------

validated_dates = pd.to_datetime(
    sales_data["order_date"],
    errors="coerce"
)

date_valid = (
    sales_data["order_date"].notna().all()
    and validated_dates.notna().all()
)

record_validation(
    "Transaction dates are present and valid",
    date_valid,
    f"{validated_dates.notna().sum()} valid dates"
)

# ------------------------------------------------------------
# SECTION 5: VALIDATE NUMERICAL BUSINESS VALUES
# ------------------------------------------------------------

# Quantity must be numeric, present, positive and integer-valued.
quantity_numeric = pd.api.types.is_numeric_dtype(
    sales_data["quantity"]
)

quantity_valid = (
    quantity_numeric
    and sales_data["quantity"].notna().all()
    and (sales_data["quantity"] > 0).all()
    and (
        sales_data["quantity"]
        % 1 == 0
    ).all()
)

record_validation(
    "Quantity values are positive integers",
    quantity_valid,
    "Quantity must be greater than zero and contain whole numbers"
)

# Unit price must be numeric, present and non-negative.
unit_price_numeric = pd.api.types.is_numeric_dtype(
    sales_data["unit_price"]
)

unit_price_valid = (
    unit_price_numeric
    and sales_data["unit_price"].notna().all()
    and (sales_data["unit_price"] >= 0).all()
)

record_validation(
    "Unit prices are numeric and non-negative",
    unit_price_valid,
    "All unit prices must be present and at least zero"
)

# Revenue must be numeric, present and non-negative.
revenue_numeric = pd.api.types.is_numeric_dtype(
    sales_data["revenue"]
)

revenue_valid = (
    revenue_numeric
    and sales_data["revenue"].notna().all()
    and (sales_data["revenue"] >= 0).all()
)

record_validation(
    "Revenue values are numeric and non-negative",
    revenue_valid,
    "All revenue values must be present and at least zero"
)

# Revenue must agree with quantity multiplied by unit price.
if quantity_numeric and unit_price_numeric and revenue_numeric:

    expected_revenue = (
        sales_data["quantity"]
        * sales_data["unit_price"]
    )

    revenue_calculation_valid = (
        (
            sales_data["revenue"]
            - expected_revenue
        ).abs() <= 0.01
    ).all()

else:
    revenue_calculation_valid = False

record_validation(
    "Revenue calculations are consistent",
    revenue_calculation_valid,
    "Revenue must equal quantity multiplied by unit price, within 0.01"
)

# ------------------------------------------------------------
# SECTION 6: VALIDATE CATEGORICAL BUSINESS VALUES
# ------------------------------------------------------------

for column in ["product", "region"]:

    values = sales_data[column].astype("string")

    # Check that all values are present.
    values_present = values.notna().all()

    # Check that no values are empty or whitespace-only.
    values_not_empty = (
        values.str.strip().ne("").fillna(False).all()
    )

    # Check that no values contain surrounding whitespace.
    values_formatted = (
        values.eq(values.str.strip()).fillna(False).all()
    )

    record_validation(
        f"{column.title()} values are present",
        values_present,
        f"{values.notna().sum()} non-missing values"
    )

    record_validation(
        f"{column.title()} values are not empty",
        values_not_empty,
        "No empty or whitespace-only values"
    )

    record_validation(
        f"{column.title()} values have consistent whitespace",
        values_formatted,
        "No leading or trailing whitespace"
    )

# ------------------------------------------------------------
# SECTION 7: VALIDATE THE DATASET STRUCTURE
# ------------------------------------------------------------

# Check identifier and quantity data types.
integer_columns = [
    "order_id",
    "customer_id",
    "quantity"
]

integer_types_valid = all(
    pd.api.types.is_integer_dtype(sales_data[column])
    for column in integer_columns
)

record_validation(
    "Identifier and quantity columns use integer data types",
    integer_types_valid,
    ", ".join(integer_columns)
)

# Check financial data types.
financial_columns = [
    "unit_price",
    "revenue"
]

financial_types_valid = all(
    pd.api.types.is_numeric_dtype(sales_data[column])
    for column in financial_columns
)

record_validation(
    "Financial columns use numeric data types",
    financial_types_valid,
    ", ".join(financial_columns)
)

# Check categorical data types.
categorical_types_valid = all(
    (
        pd.api.types.is_object_dtype(sales_data[column])
        or pd.api.types.is_string_dtype(sales_data[column])
    )
    for column in ["product", "region"]
)

record_validation(
    "Product and region columns use text-compatible data types",
    categorical_types_valid,
    "product and region"
)

# ------------------------------------------------------------
# SECTION 8: DISPLAY THE CONSOLIDATED VALIDATION REPORT
# ------------------------------------------------------------

business_rule_summary = pd.DataFrame(
    business_rule_results
)

total_rules = len(business_rule_summary)

passed_rules = (
    business_rule_summary["Result"] == "PASS"
).sum()

failed_rules = (
    business_rule_summary["Result"] == "FAIL"
).sum()

print("OVERALL BUSINESS RULE VALIDATION")
print("=" * 75)

display(business_rule_summary)

print("\nVALIDATION TOTALS")
print("=" * 75)

print(f"Total validation rules: {total_rules}")
print(f"Rules passed: {passed_rules}")
print(f"Rules failed: {failed_rules}")

# ------------------------------------------------------------
# SECTION 9: FINAL VALIDATION STATUS
# ------------------------------------------------------------

if failed_rules == 0:

    print("\nPASS: All business rule validation checks passed.")

    print(
        "PASS: The dataset satisfies the current "
        "validation requirements."
    )

else:

    print("\nFAIL: One or more business rules were not satisfied.")

    print("\nRules requiring investigation:")

    display(
        business_rule_summary.loc[
            business_rule_summary["Result"] == "FAIL"
        ]
    )

    raise AssertionError(
        "Business rule validation failed. "
        "Investigate the reported issues before proceeding."
    )

print("\nOVERALL BUSINESS RULE VALIDATION COMPLETED.")
```

## 12.6 Transaction Date Transformation

The transaction date column was identified as an object data type during the initial inspection.

Although all five date values passed the validation checks, retaining dates as text can restrict date-based operations and introduce inconsistencies during analysis.

The `order_date` column will therefore be converted into pandas datetime format.

The conversion will be applied only to the working dataset. The original source data will remain unchanged.

The transformation will be verified by checking the resulting data type, the number of valid dates and the preservation of the original date values.

```python
# ============================================================
# SECTION 12.7: CONVERT TRANSACTION DATES
# ============================================================
# PURPOSE:
# Convert the order_date column from an object data type
# into pandas datetime format.
#
# WHY THIS IS IMPORTANT:
# Datetime values support reliable date-based analysis,
# including sorting, filtering, grouping and forecasting.
#
# IMPORTANT:
# The transformation is applied only to cleaned_data.
# The original sales_data DataFrame remains unchanged.

# ------------------------------------------------------------
# SECTION 1: VERIFY THE DATE COLUMN EXISTS
# ------------------------------------------------------------

date_column = "order_date"

if date_column not in cleaned_data.columns:

    raise KeyError(
        f"Required date column not found: {date_column}"
    )

# ------------------------------------------------------------
# SECTION 2: PRESERVE ORIGINAL DATE VALUES FOR COMPARISON
# ------------------------------------------------------------

# Keep a copy of the original date values before conversion.
original_dates = sales_data[date_column].copy(deep=True)

# ------------------------------------------------------------
# SECTION 3: CONVERT THE DATE COLUMN
# ------------------------------------------------------------

try:

    cleaned_data[date_column] = pd.to_datetime(
        cleaned_data[date_column],
        errors="raise"
    )

except (ValueError, TypeError) as error:

    raise ValueError(
        "Transaction date conversion failed. "
        "Investigate the original date values."
    ) from error

# ------------------------------------------------------------
# SECTION 4: VERIFY THE RESULTING DATA TYPE
# ------------------------------------------------------------

if not pd.api.types.is_datetime64_any_dtype(
    cleaned_data[date_column]
):

    raise TypeError(
        "The transaction date column was not converted "
        "to a datetime data type."
    )

# ------------------------------------------------------------
# SECTION 5: VERIFY DATE COMPLETENESS
# ------------------------------------------------------------

valid_date_count = int(
    cleaned_data[date_column].notna().sum()
)

missing_date_count = int(
    cleaned_data[date_column].isna().sum()
)

if missing_date_count > 0:

    raise AssertionError(
        f"Date conversion produced {missing_date_count} "
        "missing date values."
    )

# ------------------------------------------------------------
# SECTION 6: VERIFY DATE VALUE PRESERVATION
# ------------------------------------------------------------

# Convert the original values temporarily for comparison.
# This checks that the conversion has not changed the
# underlying transaction dates.
expected_dates = pd.to_datetime(
    original_dates,
    errors="raise"
)

if not cleaned_data[date_column].reset_index(drop=True).equals(
    expected_dates.reset_index(drop=True)
):

    raise AssertionError(
        "The converted transaction dates do not match "
        "the original date values."
    )

# ------------------------------------------------------------
# SECTION 7: VERIFY THE ORIGINAL DATASET IS UNCHANGED
# ------------------------------------------------------------

if not sales_data[date_column].equals(original_dates):

    raise AssertionError(
        "The original transaction date column was modified."
    )

# ------------------------------------------------------------
# SECTION 8: DISPLAY TRANSFORMATION RESULTS
# ------------------------------------------------------------

print("TRANSACTION DATE TRANSFORMATION")
print("=" * 70)

print(
    "Original data type:",
    sales_data[date_column].dtype
)

print(
    "Transformed data type:",
    cleaned_data[date_column].dtype
)

print(f"Total records: {len(cleaned_data)}")
print(f"Valid dates: {valid_date_count}")
print(f"Missing dates: {missing_date_count}")

print(
    "Earliest transaction date:",
    cleaned_data[date_column].min().date()
)

print(
    "Latest transaction date:",
    cleaned_data[date_column].max().date()
)

print("\nPASS: Transaction dates converted successfully.")
print("PASS: All original date values were preserved.")
print("PASS: Original source dataset remains unchanged.")

print("\nTRANSACTION DATE TRANSFORMATION COMPLETED.")
```

Any unexpected data type will be investigated before the dataset proceeds to further preparation.

```python
# ============================================================
# SECTION 12.10: VERIFY PREPARED DATASET DATA TYPES
# ============================================================
# PURPOSE:
# Inspect and verify the data types of every column after
# the transaction date transformation.
#
# WHY THIS IS IMPORTANT:
# Analytical operations depend on columns having appropriate
# data types. Unexpected types can cause calculation errors,
# incorrect sorting or unreliable analytical results.
#
# IMPORTANT:
# This cell only inspects and verifies cleaned_data.
# It does not modify any values or data types.

# ------------------------------------------------------------
# SECTION 1: DEFINE EXPECTED DATA TYPES
# ------------------------------------------------------------

expected_integer_columns = [
    "order_id",
    "customer_id",
    "quantity"
]

expected_numeric_columns = [
    "unit_price",
    "revenue"
]

expected_text_columns = [
    "product",
    "region"
]

expected_datetime_columns = [
    "order_date"
]

# ------------------------------------------------------------
# SECTION 2: CHECK REQUIRED COLUMNS
# ------------------------------------------------------------

expected_columns = (
    expected_integer_columns
    + expected_numeric_columns
    + expected_text_columns
    + expected_datetime_columns
)

missing_columns = [
    column
    for column in expected_columns
    if column not in cleaned_data.columns
]

if missing_columns:

    raise KeyError(
        f"Required columns are missing: {missing_columns}"
    )

# ------------------------------------------------------------
# SECTION 3: VERIFY EACH COLUMN
# ------------------------------------------------------------

data_type_results = []

for column in expected_columns:

    actual_type = str(cleaned_data[column].dtype)

    if column in expected_integer_columns:

        type_valid = pd.api.types.is_integer_dtype(
            cleaned_data[column]
        )

        expected_type = "Integer"

    elif column in expected_numeric_columns:

        type_valid = pd.api.types.is_numeric_dtype(
            cleaned_data[column]
        )

        expected_type = "Numeric"

    elif column in expected_text_columns:

        type_valid = (
            pd.api.types.is_object_dtype(cleaned_data[column])
            or pd.api.types.is_string_dtype(cleaned_data[column])
        )

        expected_type = "Text-compatible"

    else:

        type_valid = pd.api.types.is_datetime64_any_dtype(
            cleaned_data[column]
        )

        expected_type = "Datetime"

    data_type_results.append({

        "Column": column,

        "Expected Type": expected_type,

        "Actual Data Type": actual_type,

        "Result": "PASS" if type_valid else "FAIL"

    })

# ------------------------------------------------------------
# SECTION 4: DISPLAY DATA TYPE ASSESSMENT
# ------------------------------------------------------------

data_type_verification = pd.DataFrame(
    data_type_results
)

print("PREPARED DATASET DATA TYPE VERIFICATION")
print("=" * 75)

display(data_type_verification)

# ------------------------------------------------------------
# SECTION 5: CALCULATE VERIFICATION TOTALS
# ------------------------------------------------------------

total_columns = len(data_type_verification)

passed_columns = (
    data_type_verification["Result"] == "PASS"
).sum()

failed_columns = (
    data_type_verification["Result"] == "FAIL"
).sum()

print("\nVERIFICATION TOTALS")
print("=" * 75)

print(f"Total columns assessed: {total_columns}")
print(f"Columns passed: {passed_columns}")
print(f"Columns failed: {failed_columns}")

# ------------------------------------------------------------
# SECTION 6: REPORT FINAL RESULT
# ------------------------------------------------------------

if failed_columns > 0:

    print("\nFAIL: Unexpected data types were identified.")

    display(
        data_type_verification.loc[
            data_type_verification["Result"] == "FAIL"
        ]
    )

    raise TypeError(
        "Prepared dataset data type verification failed."
    )

print("\nPASS: All columns have the expected data types.")
print("PASS: No unexpected data type changes detected.")

print("\nDATA TYPE VERIFICATION COMPLETED.")
```

## 12.12 Final Data Quality Checks

Following the cleaning and transformation operations, the prepared dataset will undergo a final quality assessment.

The purpose is to confirm that the transformation has not introduced new data quality issues or unintentionally altered previously validated business values.

The assessment will include:

- Missing value assessment.
- Complete duplicate record assessment.
- Numerical value assessment.
- Transaction date validity.
- Revenue calculation consistency.
- Record and column count comparison with the original dataset.

The original and prepared datasets will also be compared to confirm that the only intended transformation was the conversion of `order_date` into datetime format.

The prepared dataset will only proceed to storage if all final checks pass.

```python
# ============================================================
# SECTION 12.13: FINAL DATA QUALITY CHECKS
# ============================================================
# PURPOSE:
# Confirm that the cleaning and transformation process
# has preserved the quality and integrity of the dataset.
#
# IMPORTANT:
# The original sales_data DataFrame is used as the reference.
# The prepared cleaned_data DataFrame is assessed separately.
# No values are modified in this verification step.

# ------------------------------------------------------------
# SECTION 1: VERIFY DATASET STRUCTURE
# ------------------------------------------------------------

same_shape = sales_data.shape == cleaned_data.shape

same_columns = sales_data.columns.equals(
    cleaned_data.columns
)

if not same_shape or not same_columns:

    raise AssertionError(
        "The prepared dataset structure differs from the original."
    )

# ------------------------------------------------------------
# SECTION 2: CHECK MISSING VALUES
# ------------------------------------------------------------

missing_values = cleaned_data.isnull().sum()

total_missing_values = int(
    missing_values.sum()
)

missing_values_valid = total_missing_values == 0

# ------------------------------------------------------------
# SECTION 3: CHECK DUPLICATE RECORDS
# ------------------------------------------------------------

duplicate_records = int(
    cleaned_data.duplicated().sum()
)

duplicate_records_valid = duplicate_records == 0

# ------------------------------------------------------------
# SECTION 4: CHECK NUMERICAL VALUES
# ------------------------------------------------------------

numerical_columns = [
    "quantity",
    "unit_price",
    "revenue"
]

numerical_checks = []

for column in numerical_columns:

    values = cleaned_data[column]

    numerical_checks.append({

        "Column": column,

        "Missing Values": int(values.isna().sum()),

        "Zero Values": int((values == 0).sum()),

        "Negative Values": int((values < 0).sum()),

        "Result": (
            "PASS"
            if (
                values.notna().all()
                and (values >= 0).all()
                and (
                    (values > 0).all()
                    if column == "quantity"
                    else True
                )
            )
            else "FAIL"
        )

    })

numerical_quality_summary = pd.DataFrame(
    numerical_checks
)

numerical_values_valid = (
    numerical_quality_summary["Result"] == "PASS"
).all()

# ------------------------------------------------------------
# SECTION 5: CHECK TRANSACTION DATES
# ------------------------------------------------------------

dates_valid = (
    pd.api.types.is_datetime64_any_dtype(
        cleaned_data["order_date"]
    )
    and cleaned_data["order_date"].notna().all()
)

# ------------------------------------------------------------
# SECTION 6: CHECK REVENUE CALCULATIONS
# ------------------------------------------------------------

expected_revenue = (
    cleaned_data["quantity"]
    * cleaned_data["unit_price"]
)

revenue_differences = (
    cleaned_data["revenue"]
    - expected_revenue
).abs()

revenue_calculations_valid = (
    revenue_differences <= 0.01
).all()

# ------------------------------------------------------------
# SECTION 7: VERIFY ONLY THE INTENDED TRANSFORMATION
# ------------------------------------------------------------

# Compare all columns except order_date.
# The date column is excluded because its data type was
# intentionally changed from object to datetime64[ns].

unchanged_columns = [
    column
    for column in sales_data.columns
    if column != "order_date"
]

unchanged_values = sales_data[
    unchanged_columns
].equals(
    cleaned_data[unchanged_columns]
)

# ------------------------------------------------------------
# SECTION 8: DISPLAY QUALITY RESULTS
# ------------------------------------------------------------

final_quality_results = pd.DataFrame([

    {
        "Validation": "Dataset shape preserved",
        "Result": "PASS" if same_shape else "FAIL",
        "Details": str(cleaned_data.shape)
    },

    {
        "Validation": "Column structure preserved",
        "Result": "PASS" if same_columns else "FAIL",
        "Details": f"{len(cleaned_data.columns)} columns"
    },

    {
        "Validation": "No missing values",
        "Result": "PASS" if missing_values_valid else "FAIL",
        "Details": f"{total_missing_values} missing values"
    },

    {
        "Validation": "No duplicate records",
        "Result": "PASS" if duplicate_records_valid else "FAIL",
        "Details": f"{duplicate_records} duplicate records"
    },

    {
        "Validation": "Numerical values remain valid",
        "Result": "PASS" if numerical_values_valid else "FAIL",
        "Details": "Quantity, unit price and revenue"
    },

    {
        "Validation": "Transaction dates remain valid",
        "Result": "PASS" if dates_valid else "FAIL",
        "Details": f"{cleaned_data['order_date'].notna().sum()} valid dates"
    },

    {
        "Validation": "Revenue calculations remain consistent",
        "Result": "PASS" if revenue_calculations_valid else "FAIL",
        "Details": "Revenue equals quantity multiplied by unit price"
    },

    {
        "Validation": "Non-date values remain unchanged",
        "Result": "PASS" if unchanged_values else "FAIL",
        "Details": "All columns except order_date"
    }

])

print("FINAL DATA QUALITY ASSESSMENT")
print("=" * 80)

display(final_quality_results)

# ------------------------------------------------------------
# SECTION 9: DISPLAY NUMERICAL QUALITY DETAILS
# ------------------------------------------------------------

print("\nNUMERICAL VALUE ASSESSMENT")
print("=" * 80)

display(numerical_quality_summary)

# ------------------------------------------------------------
# SECTION 10: CALCULATE FINAL RESULTS
# ------------------------------------------------------------

total_checks = len(final_quality_results)

passed_checks = (
    final_quality_results["Result"] == "PASS"
).sum()

failed_checks = (
    final_quality_results["Result"] == "FAIL"
).sum()

print("\nFINAL VERIFICATION TOTALS")
print("=" * 80)

print(f"Total checks: {total_checks}")
print(f"Checks passed: {passed_checks}")
print(f"Checks failed: {failed_checks}")

# ------------------------------------------------------------
# SECTION 11: REPORT FINAL STATUS
# ------------------------------------------------------------

if failed_checks > 0:

    print("\nFAIL: Final data quality assessment identified issues.")

    display(
        final_quality_results.loc[
            final_quality_results["Result"] == "FAIL"
        ]
    )

    raise AssertionError(
        "Final data quality checks failed. "
        "Investigate the reported issues before storage."
    )

print("\nPASS: All final data quality checks passed.")
print("PASS: The intended date transformation was preserved.")
print("PASS: Other dataset values remain unchanged.")

print("\nFINAL DATA QUALITY ASSESSMENT COMPLETED.")
```

# 13. Processed Data Storage

## 13.1 Purpose

Following successful cleaning, transformation and final quality verification, the prepared dataset must be stored in a consistent and reproducible format.

The processed dataset will be saved separately from the original source files, using the existing `data/processed/` directory.

CSV will be used as the primary storage format because it is portable, widely supported and compatible with common data analysis tools.

The storage process will follow these principles:

- Preserve the original raw datasets.
- Store the prepared dataset separately.
- Use a clear and consistent file name.
- Avoid saving the DataFrame index as an additional column.
- Verify that the saved file exists and contains the expected records and columns.
- Reload the saved dataset to confirm that the stored information can be retrieved successfully.

The transaction date column will be interpreted as a datetime value when the processed CSV file is reloaded.

## 13.2 Prepare the Processed Data Directory

Before saving the prepared dataset, the processed data directory must be available.

The directory will be created automatically if it does not already exist.

The output file will be named `sales_data_cleaned.csv`, making its purpose clear and distinguishing it from the original raw dataset.

```python
# ============================================================
# SECTION 13.3: PREPARE THE PROCESSED DATA DIRECTORY
# ============================================================
# PURPOSE:
# Define the output location for the prepared dataset and
# ensure that the required directory exists.
#
# WHY THIS IS IMPORTANT:
# Separating raw and processed data helps preserve the
# original source files and maintain a reproducible workflow.

from pathlib import Path

# ------------------------------------------------------------
# SECTION 1: DEFINE PROJECT AND OUTPUT PATHS
# ------------------------------------------------------------

# project_dir was defined earlier when importing the
# reusable data loading module.

processed_dir = project_dir / "data" / "processed"

processed_file = processed_dir / "sales_data_cleaned.csv"

# ------------------------------------------------------------
# SECTION 2: CREATE THE DIRECTORY IF REQUIRED
# ------------------------------------------------------------

processed_dir.mkdir(
    parents=True,
    exist_ok=True
)

# ------------------------------------------------------------
# SECTION 3: VERIFY DIRECTORY CREATION
# ------------------------------------------------------------

if not processed_dir.exists():

    raise FileNotFoundError(
        "The processed data directory could not be created."
    )

if not processed_dir.is_dir():

    raise NotADirectoryError(
        "The processed data path is not a directory."
    )

# ------------------------------------------------------------
# SECTION 4: DISPLAY DIRECTORY INFORMATION
# ------------------------------------------------------------

print("PROCESSED DATA DIRECTORY")
print("=" * 70)

print(f"Project directory: {project_dir}")
print(f"Processed directory: {processed_dir}")
print(f"Output file: {processed_file.name}")

print("\nPASS: Processed data directory is available.")
print("PASS: Output file location has been defined.")

print("\nPROCESSED DATA DIRECTORY PREPARATION COMPLETED.")
```

## 13.5 Save the Prepared Dataset

The cleaned dataset will now be exported to CSV format.

The DataFrame index will be excluded because it does not represent a business data field. The existing column names and record order will be retained.

The original raw dataset will not be overwritten.

After saving, the file path, existence, file size and exported dataset dimensions will be checked.

```python
# ============================================================
# SECTION 13.6: SAVE THE PREPARED DATASET
# ============================================================
# PURPOSE:
# Export cleaned_data to the processed data directory
# in CSV format.
#
# WHY THIS IS IMPORTANT:
# Saving the prepared dataset creates a reusable output
# that can be loaded by subsequent analytical modules.
#
# IMPORTANT:
# The original raw files will not be overwritten.
# The DataFrame index will not be exported as a data column.

# ------------------------------------------------------------
# SECTION 1: VERIFY THE DATASET BEFORE EXPORT
# ------------------------------------------------------------

if not isinstance(cleaned_data, pd.DataFrame):

    raise TypeError(
        "The prepared dataset must be a pandas DataFrame."
    )

if cleaned_data.empty:

    raise ValueError(
        "The prepared dataset contains no records."
    )

if not processed_dir.exists():

    raise FileNotFoundError(
        "The processed data directory does not exist."
    )

# ------------------------------------------------------------
# SECTION 2: EXPORT THE DATASET
# ------------------------------------------------------------

try:

    cleaned_data.to_csv(
        processed_file,
        index=False,
        encoding="utf-8"
    )

except (OSError, PermissionError, ValueError) as error:

    raise IOError(
        "The prepared dataset could not be saved."
    ) from error

# ------------------------------------------------------------
# SECTION 3: VERIFY THE EXPORTED FILE
# ------------------------------------------------------------

if not processed_file.exists():

    raise FileNotFoundError(
        "The CSV file was not created successfully."
    )

file_size_bytes = processed_file.stat().st_size

if file_size_bytes == 0:

    raise ValueError(
        "The exported CSV file is empty."
    )

# ------------------------------------------------------------
# SECTION 4: DISPLAY EXPORT RESULTS
# ------------------------------------------------------------

print("PROCESSED DATA EXPORT")
print("=" * 70)

print(f"File name: {processed_file.name}")
print(f"File location: {processed_file}")
print(f"Records exported: {len(cleaned_data)}")
print(f"Columns exported: {len(cleaned_data.columns)}")
print(f"File size: {file_size_bytes:,} bytes")

print("\nPASS: Prepared dataset exported successfully.")
print("PASS: CSV file exists and is not empty.")
print("PASS: Original raw dataset was not overwritten.")

print("\nPROCESSED DATA EXPORT COMPLETED.")
```

## 13.8 Reload and Verify the Processed Dataset

The exported CSV file will now be loaded into a new pandas DataFrame.

This independent reload verifies that the saved file can be accessed and interpreted correctly.

The transaction date column will be parsed as a datetime value during loading because CSV files do not preserve pandas data types automatically.

The reloaded dataset will then be compared with the original `cleaned_data` DataFrame.

The comparison will verify:

- Record and column counts.
- Column names and order.
- Data values.
- Data types.
- Transaction date preservation.
- Absence of additional index columns.

The processed dataset will only be considered successfully stored once these checks have passed.

```python
# ============================================================
# SECTION 13.9: RELOAD AND VERIFY THE PROCESSED DATASET
# ============================================================
# PURPOSE:
# Independently reload the exported CSV file and compare it
# with the prepared cleaned_data DataFrame.
#
# WHY THIS IS IMPORTANT:
# A successful export does not automatically guarantee that
# the saved data can be reloaded without unexpected changes.
#
# IMPORTANT:
# The reloaded dataset will be stored in a separate variable
# named reloaded_data. The original DataFrames remain intact.

# ------------------------------------------------------------
# SECTION 1: VERIFY THAT THE EXPORTED FILE EXISTS
# ------------------------------------------------------------

if not processed_file.exists():

    raise FileNotFoundError(
        f"Processed dataset not found: {processed_file}"
    )

# ------------------------------------------------------------
# SECTION 2: LOAD THE EXPORTED CSV FILE
# ------------------------------------------------------------

try:

    reloaded_data = pd.read_csv(
        processed_file,
        parse_dates=["order_date"],
        encoding="utf-8"
    )

except (OSError, ValueError, pd.errors.ParserError) as error:

    raise ValueError(
        "The processed CSV file could not be reloaded."
    ) from error

# ------------------------------------------------------------
# SECTION 3: VERIFY DATASET DIMENSIONS
# ------------------------------------------------------------

shape_matches = (
    reloaded_data.shape == cleaned_data.shape
)

# ------------------------------------------------------------
# SECTION 4: VERIFY COLUMN NAMES AND ORDER
# ------------------------------------------------------------

columns_match = (
    reloaded_data.columns.equals(cleaned_data.columns)
)

# ------------------------------------------------------------
# SECTION 5: VERIFY DATA VALUES AND DATA TYPES
# ------------------------------------------------------------

# This comparison checks the complete DataFrames, including
# their column names, values, row order and data types.
#
# check_dtype=True requires matching data types.
# check_exact=True requires exact numerical value equality.

try:

    pd.testing.assert_frame_equal(
        reloaded_data,
        cleaned_data,
        check_dtype=True,
        check_exact=True
    )

    data_matches = True

except AssertionError:

    data_matches = False

# ------------------------------------------------------------
# SECTION 6: VERIFY THE TRANSACTION DATE DATA TYPE
# ------------------------------------------------------------

date_type_valid = (
    pd.api.types.is_datetime64_any_dtype(
        reloaded_data["order_date"]
    )
)

# ------------------------------------------------------------
# SECTION 7: CHECK FOR AN UNEXPECTED INDEX COLUMN
# ------------------------------------------------------------

unexpected_index_columns = [
    column
    for column in reloaded_data.columns
    if column.lower() in ["index", "unnamed: 0"]
]

index_column_valid = (
    len(unexpected_index_columns) == 0
)

# ------------------------------------------------------------
# SECTION 8: DISPLAY RELOADING RESULTS
# ------------------------------------------------------------

reload_results = pd.DataFrame([

    {
        "Verification": "Dataset dimensions match",
        "Result": "PASS" if shape_matches else "FAIL",
        "Details": str(reloaded_data.shape)
    },

    {
        "Verification": "Column names and order match",
        "Result": "PASS" if columns_match else "FAIL",
        "Details": f"{len(reloaded_data.columns)} columns"
    },

    {
        "Verification": "Data values and types match",
        "Result": "PASS" if data_matches else "FAIL",
        "Details": "Complete DataFrame comparison"
    },

    {
        "Verification": "Transaction date type preserved",
        "Result": "PASS" if date_type_valid else "FAIL",
        "Details": str(reloaded_data["order_date"].dtype)
    },

    {
        "Verification": "No unexpected index column",
        "Result": "PASS" if index_column_valid else "FAIL",
        "Details": "No additional index column detected"
    }

])

print("PROCESSED DATA RELOAD VERIFICATION")
print("=" * 75)

display(reload_results)

# ------------------------------------------------------------
# SECTION 9: CALCULATE FINAL VERIFICATION TOTALS
# ------------------------------------------------------------

total_checks = len(reload_results)

passed_checks = (
    reload_results["Result"] == "PASS"
).sum()

failed_checks = (
    reload_results["Result"] == "FAIL"
).sum()

print("\nRELOAD VERIFICATION TOTALS")
print("=" * 75)

print(f"Total checks: {total_checks}")
print(f"Checks passed: {passed_checks}")
print(f"Checks failed: {failed_checks}")

# ------------------------------------------------------------
# SECTION 10: REPORT FINAL RESULT
# ------------------------------------------------------------

if failed_checks > 0:

    print("\nFAIL: Processed dataset verification identified issues.")

    display(
        reload_results.loc[
            reload_results["Result"] == "FAIL"
        ]
    )

    raise AssertionError(
        "The exported dataset does not fully match "
        "the prepared DataFrame."
    )

print("\nPASS: Processed dataset reloaded successfully.")
print("PASS: All data values and data types were preserved.")
print("PASS: No unexpected index column was introduced.")

print("\nPROCESSED DATA RELOAD VERIFICATION COMPLETED.")
```
